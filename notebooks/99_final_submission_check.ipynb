## 3. اقرأ لقطة المشروع

إذا تركت بيانات GitHub فارغة، اختر ZIP من جهازك. لا تدخل كلمة مرور أو token. يتحقق القارئ من أسماء المسارات وحدود حجم الحزمة قبل فكها.

```python
from submission_contract import extract_project

if PROJECT_REPOSITORY or PROJECT_SHA:
    if not re.fullmatch(r"[A-Za-z0-9_.-]+/[A-Za-z0-9_.-]+", PROJECT_REPOSITORY) or not re.fullmatch(r"[0-9a-fA-F]{40}", PROJECT_SHA):
        raise ValueError("Enter owner/repository and the full 40-character SHA.")
    archive = CHECK_HOME / "source.zip"
    url = f"https://codeload.github.com/{PROJECT_REPOSITORY}/zip/{PROJECT_SHA}"
    with urllib.request.urlopen(url, timeout=60) as response:
        payload = response.read(100*1024*1024+1)
    if len(payload)>100*1024*1024:
        raise ValueError("Archive exceeds 100 MiB.")
    archive.write_bytes(payload)
else:
    if not PROJECT_ZIP:
        from google.colab import files
        uploaded = files.upload()
        if len(uploaded)!=1:
            raise ValueError("Choose exactly one project ZIP.")
        PROJECT_ZIP = next(iter(uploaded))
    archive = Path(PROJECT_ZIP)
PROJECT = extract_project(archive, CHECK_HOME / "project")
print("Project:", PROJECT)
print("Archive SHA256:", hashlib.sha256(archive.read_bytes()).hexdigest())
```
